### \d - Denotes any Digit

```python
import re
text =''' 
You can reach me at 123-4562-7890 or email me at example@email.com'''
pattern = r'\d{3}-\d{4}-\d{4}'
matches = re.findall(pattern, text)
matches
```

### | - Donates multiples searches 

```python
import re
text= '''
Scan me using Cnic 35401-9546-658-9 or scan by number 0335-4566125'''
pattern =r'\d{5}-\d{4}-\d{3}-\d{1}|\d{4}-\d{7}'
matches=re.findall(pattern,text)
print(matches)
```

### USING ( ) 

```python
import re
text ='''
My cousin planned his wedding on Thursday at 13 December 2024 at (09):(55):(01)
'''
pattern = r'\(\d{2}\):\(\d{2}\):\(\d{2}\)' 
matches = re.findall(pattern, text)
print(matches)
```
